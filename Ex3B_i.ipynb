## AIM

To build and validate a Linear Regression model on the UCI Diabetes dataset.

## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import pandas, numpy, sklearn.datasets and sklearn.linear_model.

**Step 3:** Load the UCI Diabetes dataset using load_diabetes(return_X_y=True).

**Step 4:** Select BMI (column index 2) as the single feature.

**Step 5:** Split the data into training (first 422 rows) and testing (last 20 rows).

**Step 6:** Create a LinearRegression() object.

**Step 7:** Fit the model with training data using fit().

**Step 8:** Predict the target values on the test set using predict().

**Step 9:** Calculate the Coefficient, Mean Squared Error and R² Score.

**Step 10:** Plot the test data and regression line using matplotlib.

**Step 11:** Stop the program.



## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model
from sklearn.metrics import mean_squared_error, r2_score

diabetes_X, diabetes_y = datasets.load_diabetes(return_X_y=True)  # built-in UCI diabetes data
diabetes_X = diabetes_X[:, np.newaxis, 2]                     # take only column 2 (BMI)

diabetes_X_train = diabetes_X[:-20]                           # all rows except last 20: training
diabetes_X_test  = diabetes_X[-20:]                           # last 20 rows: testing
diabetes_y_train = diabetes_y[:-20]
diabetes_y_test  = diabetes_y[-20:]

regr = linear_model.LinearRegression()
regr.fit(diabetes_X_train, diabetes_y_train)                  # train the model
y_pred = regr.predict(diabetes_X_test)                        # predict for test data

print("Coefficient:", regr.coef_)                             # slope of the line
print("Mean squared error: %.2f" % mean_squared_error(diabetes_y_test, y_pred))
print("Coefficient of determination (R²): %.2f" % r2_score(diabetes_y_test, y_pred))  # R² near 1 = good fit

plt.scatter(diabetes_X_test, diabetes_y_test, color='blue', label='Actual')
plt.plot(diabetes_X_test, y_pred, color='red', linewidth=3, label='Predicted')
plt.xlabel('BMI (scaled)'); plt.ylabel('Disease progression')
plt.title('Linear Regression - Diabetes Dataset')
plt.legend(); plt.grid(True); plt.show()
print("Intercept:", regr.intercept_)
```
