## ALGORITHM

**Step 1:** Start the program.

**Step 2:** Import pandas as pd and matplotlib.pyplot as plt.

**Step 3:** Enable inline plotting using %matplotlib inline.

**Step 4:** Create a DataFrame 'sales' with Category, Sales and Profit of five categories of SuperMart Stores.

**Step 5:** Display the sales data.

**Step 6:** Draw a bar chart of Sales by Category using plt.bar() with title and axis labels.

**Step 7:** Draw a donut chart of Profit share by Category using plt.pie() with wedgeprops width 0.4.

## PROGRAM

_Click the code and press **Shift + Enter** to run it. The output appears just below the code._

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt

# Sample sales data (what Power BI dashboard displays)
sales = pd.DataFrame({
    'Category': ['Electronics', 'Clothing', 'Grocery', 'Furniture', 'Books'],
    'Sales'   : [250000, 180000, 400000, 140000,  90000],
    'Profit'  : [ 45000,  30000,  60000,  22000,  18000]
})
print("Sample Sales Data (SuperMart Stores):")
print(sales)

# Bar chart - Sales by Category
plt.figure(figsize=(9, 5))
plt.bar(sales['Category'], sales['Sales'], color='orange', edgecolor='black')  # bar chart like Power BI
plt.title('Sales by Category (Power BI style)')
plt.xlabel('Category'); plt.ylabel('Sales (Rupees)')
plt.grid(True, axis='y'); plt.show()

# Donut chart - Profit Share
plt.figure(figsize=(7, 7))
plt.pie(sales['Profit'], labels=sales['Category'], autopct='%1.1f%%',
        wedgeprops=dict(width=0.4))                           # ring width makes it a donut
plt.title('Profit Share by Category (Donut Chart)')
plt.show()
```
